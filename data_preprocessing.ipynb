```python
import re

def remove_repeated_sentences(text):
    sentences = text.split("\n")

    unique_sentences = []
    seen_sentences = set()

    for sentence in sentences:
        clean_sentence = sentence.strip()
        if clean_sentence and clean_sentence not in seen_sentences:
            unique_sentences.append(clean_sentence)
            seen_sentences.add(clean_sentence)

    return "\n".join(unique_sentences)

df_no_dupl = df.copy()
df_no_dupl["text"] = df_no_dupl["text"].apply(remove_repeated_sentences)
```

```python
def split_text(row):
    pattern_established = r"у\s*с\s*т\s*а\s*н\s*о\s*в\s*и\s*л\s*(?:\n|\:\n)"
    pattern_sentenced = r"п\s*р\s*и\s*г\s*о\s*в\s*о\s*р\s*и\s*л\s*(?:\n|\:\n)"
    
    text = row["text"]

    established_match = re.search(pattern_established, text, re.IGNORECASE)
    established_i = established_match.start()
    sentenced_match= re.search(pattern_sentenced, text, re.IGNORECASE)
    sentenced_i = sentenced_match.start()

    part1 = text[:established_i]
    part2 = text[established_i:sentenced_i]
    part3 = text[sentenced_i:]

    return pd.Series([part1, part2, part3])

df[["preamble", "description", "sentence"]] = df.apply(split_text, axis=1)
```
