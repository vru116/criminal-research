# Оставляем нужное

```python
import pandas as pd

df = pd.read_csv('parsed_data.csv')
```

```python
df.shape
```

```python
df = df[['id', 'region', 'entryDate', 'names', 'judge', 'decision', 'accused', 'articles', 'link_text', 'text']]
df.head()
```

```python
df["entryDate"]
```

```python
df.isna().sum()
```

```python
df.loc[:, "judge"] = df["judge"].fillna("Неизвестно")
df.loc[:, "accused"] = df["accused"].fillna("Неизвестно")
```

```python
df.isna().sum()
```

```python
import re

pattern_established = r"у\s*с\s*т\s*а\s*н\s*о\s*в\s*и\s*л"
pattern_sentenced = r"п\s*р\s*и\s*г\s*о\s*в\s*о\s*р\s*и\s*л"

df = df[df['text'].str.contains(pattern_established, regex=True, flags=re.IGNORECASE) & 
        df['text'].str.contains(pattern_sentenced, regex=True, flags=re.IGNORECASE)]

print(df.shape)
```

```python
def split_text(row):
    text = row["text"]

    established_match = re.search(pattern_established, text, re.IGNORECASE)
    established_i = established_match.start()
    sentenced_match= re.search(pattern_sentenced, text, re.IGNORECASE)
    sentenced_i = sentenced_match.start()

    part1 = text[:established_i]
    part2 = text[established_i:sentenced_i]
    part3 = text[sentenced_i:]

    return pd.Series([part1, part2, part3])

df[["preamble", "description", "sentence"]] = df.apply(split_text, axis=1)
```
