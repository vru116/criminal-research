```python
def clean_value(value, column):
    if isinstance(value, list):
        return value
    if pd.isna(value):
        print(value, column)
        return "неизвестно"
    else:
        value = value.replace("\n", "")
        if column != "crime_region":
            value = value.replace("-", ",")
        value = value.replace("о́", "о")
        value = value.replace("конфликт, личная неприязнь", "личная неприязнь")
        value = value.replace("конфликт", "личная неприязнь")
        value = value.replace("незизвестно", "неизвестно")
        value = value.replace("неизестно", "неизвестно")
        value = value.replace("умышленное (не убил)", "умышленное")

        if "неизвестно" in value:
            value = value.strip()

        if column in ["prior_convictions", "drugs"]:
            if value == "неизвестно":
                value = "нет"
        if column in ["relationship_between_accused_and_victim", "time_of_day", "day_of_week", "season", "year", "crime_region", "motive"]:
            if value == "нет":
                value = "неизвестно"


        if value == "1990":
            value = "27"

        if value == "республика Татарстан":
            value = "Республика Татарстан"

        if "," in value:
            values = [v.strip() for v in value.split(",")]
            return values
        return value

for col in ['gender_accused', 'age_accused', 'alcohol', 'drugs', 'mental_disorder', 'prior_convictions', 'gender_victim', 'age_victim', 'relationship_between_accused_and_victim', 'time_of_day', 'day_of_week', 'season', 'year', 'crime_region', 'location', 'method', 'cruelty', 'motive', 'witnesses', 'precrime_argument', 'prison_term', 'intentional']:
    df_with_marking[col] = df_with_marking[col].apply(lambda x: clean_value(x, col))
```

```python
def clean_value(value, column):
    if column == 'prison_term':
        if value != "неизвестно":
            return float(value)
    if column == "year":
        if value != "неизвестно":
            return int(float(value))
for col in ['year', 'prison_term']:
    df_with_marking[col] = df_with_marking[col].apply(lambda x: clean_value(x, col))
```
