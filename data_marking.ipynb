```python
import pandas as pd

df = pd.read_csv('preprocessed_data.csv')
```

```python
df.head()
```

```python
df = df.assign(
    gender_accused="", age_accused="", alcohol="", drugs="", mental_disorder="", prior_convictions="",
    gender_victim="", age_victim="", relationship_between_accused_and_victim="",
    time_of_day="", day_of_week="", season="", year="",
    crime_region="", location="", method="", cruelty="", intentional="", motive="",
    witnesses="", precrime_argument="", prison_term=""
)

df.head()
```

```python
df_sample = df.sample(n=100, random_state=42)
df_sample
```

```python
import numpy as np

df_sample1, df_sample2, df_sample3, df_sample4 = np.array_split(df_sample, 4)
```

```python
df_sample1.shape, df_sample2.shape, df_sample3.shape, df_sample4.shape
```

```python
df_sample1.to_csv('df_sample1.csv', index=False)
df_sample2.to_csv('df_sample2.csv', index=False)
df_sample3.to_csv('df_sample3.csv', index=False)
df_sample4.to_csv('df_sample4.csv', index=False)
```
